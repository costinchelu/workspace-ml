# Polynomial Regression

Polynomial Linear Regression models non-linear relationships between an independent variable $x$ and a dependent variable $\hat{y}$.

Instead of fitting a straight line, it models curves by adding powers of the predictor variable (e.g., $x^2$, $x^3$). It is still considered linear regression because the model is linear in terms of its parameters (coefficients $\beta$), even though the boundary curve is non-linear relative to $x$.

For a polynomial of degree $n$, the model equation is:

$$\hat{y} = \beta_0 + \beta_1 x + \beta_2 x^2 + \beta_3 x^3 + \dots + \beta_n x^n + \varepsilon$$

Where:

- $\hat{y}$ = Target (dependent variable)
- $x$ = Feature (independent variable)
- $\beta_0$ = y-intercept
- $\beta_1$, $\beta_2$, $\dots$, $\beta_n$ = Coefficients corresponding to each degree power of $x$
- $n$ = Degree of the polynomial (e.g., $n=2$ for a quadratic curve, $n=3$ for a cubic curve)
- $\varepsilon$ = Residual error term

## Importing the libraries

```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```
dataset = pd.read_csv('data/Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

## Training the Linear Regression model on the whole dataset

```
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

## Training the Polynomial Regression model on the whole dataset

```
from sklearn.preprocessing import PolynomialFeatures

# for 2nd degree polynomial regression:
poly_reg_2 = PolynomialFeatures(degree = 2)
X_poly_2 = poly_reg_2.fit_transform(X)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly_2, y)

# for 3rd degree polynomial regression:
poly_reg_3 = PolynomialFeatures(degree = 3)
X_poly_3 = poly_reg_3.fit_transform(X)
lin_reg_3 = LinearRegression()
lin_reg_3.fit(X_poly_3, y)

# for 4th degree polynomial regression:
poly_reg_4 = PolynomialFeatures(degree = 4)
X_poly_4 = poly_reg_4.fit_transform(X)
lin_reg_4 = LinearRegression()
lin_reg_4.fit(X_poly_4, y)
```

## Visualising the Linear Regression results

```
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_reg.predict(X), color = 'blue')
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

## Visualising the Polynomial Regression results (2nd degree polynomial)

```
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_reg_2.predict(X_poly_2), color = 'blue')
plt.title('Truth or Bluff (2nd degree polynomial regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

## Visualising the Polynomial Regression results (3rd degree polynomial)

```
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_reg_3.predict(X_poly_3), color = 'blue')
plt.title('Truth or Bluff (3rd degree polynomial regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

## Visualising the Polynomial Regression results (4th degree polynomial)

```
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_reg_4.predict(X_poly_4), color = 'blue')
plt.title('Truth or Bluff (4th degree polynomial regression)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

## Visualising the Polynomial Regression results (for higher resolution and smoother curve)

```
# Create closely spaced, column-shaped inputs (0,1 precision for X axis) to plot a smooth degree-4 curve.
X_grid = np.arange(X.min(), X.max(), 0.1)
X_grid = X_grid.reshape((len(X_grid), 1))

X_poly_4_smooth = poly_reg_4.fit_transform(X_grid)

plt.scatter(X, y, color = 'red')
plt.plot(X_grid, lin_reg_4.predict(X_poly_4_smooth), color = 'blue')
plt.title('Truth or Bluff (4th degree polynomial regression - high resolution)')
plt.xlabel('Position level')
plt.ylabel('Salary')
plt.show()
```

## Predicting a new result with Linear Regression

```
# considering a grade ~ 6.5
grade = [[6.5]] # 1x1 matrix

lin_reg.predict(grade)
# the predicted salary ~ 330,378 is unusually large, which is a sign of underfitting. The model is not capturing the non-linear relationship between position level and salary.
```

## Predicting a new result with Polynomial Regression

```
# considering a grade ~ 6.5
lin_reg_4.predict(poly_reg_4.fit_transform(grade))
# the predicted salary ~ 158,862 is more reasonable and aligns with the non-linear trend observed in the data. This indicates that the polynomial regression model is better suited for capturing the relationship between position level and salary.
```
