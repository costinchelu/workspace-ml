# Kernel PCA

## Introduction to PCA

**Principal Component Analysis (PCA)** is an unsupervised dimensionality reduction technique that projects data onto orthogonal directions called principal components. These directions capture the greatest variance, allowing us to retain much of the data's variation with fewer features. Standard PCA is linear; **Kernel PCA** uses a kernel function to perform PCA implicitly in a transformed feature space, capturing nonlinear structure in the original data.

### Why use Kernel PCA?

Linear PCA can summarize data that lies near a line or plane, but it may miss curved patterns. For example, points arranged in concentric circles cannot be separated by a straight line in their original two-dimensional space. A nonlinear feature mapping can reveal structure that a linear projection would miss. Kernel PCA looks for high-variance directions in that mapped space, then represents each sample using a smaller number of coordinates.

### The kernel trick

Rather than explicitly constructing a potentially very high-dimensional mapping $\phi(x)$, a kernel computes inner products in that space directly: $$k(x_i, x_j) = \langle \phi(x_i), \phi(x_j) \rangle$$

Kernel PCA builds the kernel matrix $K_{ij} = k(x_i, x_j)$ for the training samples, centers it in feature space, and uses its leading eigenvectors and eigenvalues to obtain the reduced coordinates. New samples are projected using their kernel similarities to the training samples.

This notebook uses the **RBF kernel**:

$$k(x_i, x_j) = \exp\left(-\gamma \lVert x_i - x_j \rVert^2\right), \qquad \gamma > 0.$$

Nearby samples have similarity close to 1, while distant samples have similarity close to 0. The parameter $\gamma$ controls the neighborhood scale: smaller values give a broader notion of similarity, while larger values make similarity more local. Here, `gamma` is left at scikit-learn's default of $1 / n_{\text{features}}$; choosing it carefully can substantially affect the embedding.

### Practical considerations

- **Scale features first:** the RBF kernel depends on distances, so large-scale features can otherwise dominate. Fit both the scaler and Kernel PCA on training data only to avoid data leakage.
- **Unsupervised does not mean class-separating:** Kernel PCA does not use class labels. A useful nonlinear embedding may help a classifier, but better accuracy or perfect separation is not guaranteed. Logistic Regression remains linear in the reduced coordinates, although its boundaries can be nonlinear in the original feature space.
- **Two components are a visualization choice:** they make decision regions easy to plot but may discard useful information. Select the kernel, `gamma`, and component count using validation or cross-validation within the training data, keeping the test set for final evaluation.
- **Interpretation and cost differ from PCA:** kernel components are not simple linear combinations of the original features, and scikit-learn's `KernelPCA` does not provide PCA's `explained_variance_ratio_`. A full kernel matrix requires $O(n^2)$ memory for $n$ training samples, making large datasets more expensive to process.

## About this notebook

This notebook applies **Kernel PCA with an RBF (radial basis function) kernel** to the Wine dataset:

1. Split the data into training and test sets, then standardize the features.
2. Reduce the features to **two kernel principal components**, fitting the transformation on the training set only.
3. Train a **Logistic Regression** classifier on the reduced training data.
4. Evaluate test predictions using a **confusion matrix** and **accuracy**, then visualize the training and test decision regions.

The goal is to illustrate how nonlinear dimensionality reduction can be combined with classification and two-dimensional visualization.

```
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```
dataset = pd.read_csv('../data/Wine.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

## Splitting the dataset into the Training set and Test set

```
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

## Feature Scaling

```
from sklearn.preprocessing import StandardScaler

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## Applying Kernel PCA

```
from sklearn.decomposition import KernelPCA

kpca = KernelPCA(n_components=2, kernel='rbf')
X_train = kpca.fit_transform(X_train)
X_test = kpca.transform(X_test)
```

## Training the Logistic Regression model on the Training set

```
from sklearn.linear_model import LogisticRegression

classifier = LogisticRegression(random_state=0)
classifier.fit(X_train, y_train)
```

## Making the Confusion Matrix

```
from sklearn.metrics import confusion_matrix, accuracy_score

y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

## Visualising the Training set results

```
from matplotlib.colors import ListedColormap

X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(
    np.arange(
        start=X_set[:, 0].min() - 1,
        stop=X_set[:, 0].max() + 1,
        step=0.01
    ),
    np.arange(
        start=X_set[:, 1].min() - 1,
        stop=X_set[:, 1].max() + 1,
        step=0.01
    ))
plt.contourf(
    X1,
    X2,
    classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
    alpha=0.75,
    cmap=ListedColormap(('red', 'green', 'blue'))
)
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(
        X_set[y_set == j, 0],
        X_set[y_set == j, 1],
        c=ListedColormap(('red', 'green', 'blue'))(i),
        label=j)
plt.title('Logistic Regression (Training set)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```

## Visualising the Test set results

```
from matplotlib.colors import ListedColormap

X_set, y_set = X_test, y_test
X_set, y_set = X_test, y_test
X1, X2 = np.meshgrid(
    np.arange(
        start=X_set[:, 0].min() - 1,
        stop=X_set[:, 0].max() + 1,
        step=0.01
    ), np.arange(
        start=X_set[:, 1].min() - 1,
        stop=X_set[:, 1].max() + 1,
        step=0.01
    ))
plt.contourf(X1,
             X2,
             classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha=0.75,
             cmap=ListedColormap(('red', 'green', 'blue')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(
        X_set[y_set == j, 0],
        X_set[y_set == j, 1],
        c=ListedColormap(('red', 'green', 'blue'))(i),
        label=j)
plt.title('Logistic Regression (Test set)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```
