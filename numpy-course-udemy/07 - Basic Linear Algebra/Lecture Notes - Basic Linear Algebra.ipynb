<b>Dot Product</b>

Multiplies matching components and adds the products:

$$
\begin{aligned}
\operatorname{dot}(\mathbf{first\_vector}, \mathbf{second\_vector})
&= \begin{bmatrix} 1 & 2 & 4 \end{bmatrix} \begin{bmatrix} 2 \\ -1 \\ 9 \end{bmatrix} \\n&= (1 \cdot 2) + (2 \cdot -1) + (4 \cdot 9) \\n&= 2 - 2 + 36 \\n&= 36
\end{aligned}
$$

```python
np.dot(first_vector, second_vector)
```

```python
def are_orthogonal(vect1, vect2):
    return np.dot(vect1, vect2) == 0
```

```python
are_orthogonal(first_vector, second_vector)
```

## Cross Product and Length


<b>Cross Product</b>

For two three-dimensional vectors, `np.cross` returns a vector perpendicular to both inputs. Its direction follows the right-hand rule, and its magnitude equals the area of the parallelogram formed by the vectors.

```python
np.cross(first_vector, second_vector)
```

```python
np.cross(np.arange(2), np.arange(1,3))
```

For `first_vector = [1, 2, 4]` and `second_vector = [2, -1, 9]`:

$$
\begin{aligned}
\operatorname{cross}(\mathbf{first\_vector}, \mathbf{second\_vector})
&= \begin{vmatrix} \mathbf{i} & \mathbf{j} & \mathbf{k} \\ 1 & 2 & 4 \\ 2 & -1 & 9 \end{vmatrix} \\n&= \mathbf{i}(2 \cdot 9 - 4 \cdot -1) - \mathbf{j}(1 \cdot 9 - 4 \cdot 2) + \mathbf{k}(1 \cdot -1 - 2 \cdot 2) \\n&= 22\mathbf{i} - \mathbf{j} - 5\mathbf{k} \\n&= [22, -1, -5]
\end{aligned}
$$

```python
cross = np.cross(first_vector, second_vector)
```

```python
np.dot(first_vector, cross)
```

```python
np.dot(second_vector, cross)
```

```python
def angle(vect1, vect2):
    cos_angle = np.dot(vect1, vect2) / (linalg.norm(vect1) * linalg.norm(vect2))
    return np.arccos(cos_angle)
```

<b>Matrix Product</b>

### Matrix to vector

To multiply a 4-dimensional vector by a $3 \times 4$ matrix, we must use a **column vector** of size $4 \times 1$.

$$A = \begin{bmatrix} 1 & 0 & 2 & 1 \\ 0 & 1 & -1 & 3 \\ 2 & 1 & 0 & 1 \end{bmatrix}$$
$$\mathbf{v} = \begin{bmatrix} 1 \\ 2 \\ 3 \\ 4 \end{bmatrix}$$

$$(3 \times \mathbf{4}) \times (\mathbf{4} \times 1) \longrightarrow \text{Valid! Inner dimensions match. Result is a } 3 \times 1 \text{ vector.}$$
We take the dot product of each row of the matrix with the vector:
1. **Row 1:** $(1 \cdot 1) + (0 \cdot 2) + (2 \cdot 3) + (1 \cdot 4) = 1 + 0 + 6 + 4 = \mathbf{11}$
2. **Row 2:** $(0 \cdot 1) + (1 \cdot 2) + (-1 \cdot 3) + (3 \cdot 4) = 0 + 2 - 3 + 12 = \mathbf{11}$
3. **Row 3:** $(2 \cdot 1) + (1 \cdot 2) + (0 \cdot 3) + (1 \cdot 4) = 2 + 2 + 0 + 4 = \mathbf{8}$

$$\begin{bmatrix} 1 & 0 & 2 & 1 \\ 0 & 1 & -1 & 3 \\ 2 & 1 & 0 & 1 \end{bmatrix} \begin{bmatrix} 1 \\ 2 \\ 3 \\ 4 \end{bmatrix} = \begin{bmatrix} 11 \\ 11 \\ 8 \end{bmatrix}$$

```python
first_matrix @ first_vector
```

### Matrix to matrix

$$\text{Matrix }A=\begin{bmatrix}\color{#3271EA}{1}&\color{#3271EA}{2}&\color{#3271EA}{3}\\ \color{#DB372D}{4}&\color{#DB372D}{5}&\color{#DB372D}{6}\end{bmatrix},\quad \text{Matrix }B=\begin{bmatrix}7&8\\ 9&1\\ 2&3\end{bmatrix}$$

$$\text{Result size: } (2 \times 3) \times (3 \times 2) = 2 \times 2$$


Step-by-Step Calculation:

| Position in Result | Operation (Row of $A \cdot$ Column of $B$) | Calculation | Result |
| :--- | :--- | :--- | :--- |
| Top-Left (Row 1, Col 1) | Row 1 ($\color{#3271EA}{1,2,3}$) $\times$ Col 1 ($7, 9, 2$) | $(1 \times 7) + (2 \times 9) + (3 \times 2) = 7 + 18 + 6$ | 31 |
| Top-Right (Row 1, Col 2) | Row 1 ($\color{#3271EA}{1,2,3}$) $\times$ Col 2 ($8, 1, 3$) | $(1 \times 8) + (2 \times 1) + (3 \times 3) = 8 + 2 + 9$ | 19 |
| Bottom-Left (Row 2, Col 1) | Row 2 ($\color{#DB372D}{4,5,6}$) $\times$ Col 1 ($7, 9, 2$) | $(4 \times 7) + (5 \times 9) + (6 \times 2) = 28 + 45 + 12$ | 85 |
| Bottom-Right (Row 2, Col 2) | Row 2 ($\color{#DB372D}{4,5,6}$) $\times$ Col 2 ($8, 1, 3$) | $(4 \times 8) + (5 \times 1) + (6 \times 3) = 32 + 5 + 18$ | 55 |

Final Result:

$$A \times B = \begin{bmatrix}\mathbf{31} & \mathbf{19} \\ \mathbf{85} & \mathbf{55}\end{bmatrix}$$

## Solving Linear Systems

<b>Matrix Determinant</b>

```python
linalg.det(first_matrix)
```

```python
linalg.det(second_matrix)
```

```python
non_invertible = np.array([[1, 2, 0], [-2, 10, 3], [-3, 8, 3]])
```

```python
# need to round as determinant is not returning an exact 0
np.around(linalg.det(non_invertible), 10)
```

<b>Inverse Matrices</b>

```python
# should return identity matrix
np.around(linalg.inv(first_matrix) @ first_matrix, 10)
```

```python
linalg.inv(non_invertible)
```

<b>Solving Linear Systems</b>

Problem which have one unique solution.

```python
linalg.solve(first_matrix, first_vector)
```

```python
# checking the solution
first_matrix @ linalg.solve(first_matrix, first_vector)
```

Problem with no solutions.

```python
# very big values - sign that the linear system doesn't have any solutions
linalg.solve(non_invertible, first_vector)
```

```python
non_invertible @ linalg.solve(non_invertible, first_vector)
```

Problem with infinitely many solutions.

```python
third_vector = np.array([1, -2, -3])
```

```python
linalg.solve(non_invertible, third_vector)
```

```python
non_invertible @ linalg.solve(non_invertible, third_vector)
```
