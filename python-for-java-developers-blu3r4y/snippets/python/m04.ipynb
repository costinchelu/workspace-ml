# Comparison of Division: Java vs. Python

While both Java and Python handle basic arithmetic intuitively for positive numbers, they differ significantly in operator semantics, rounding behavior for negative numbers, modulo definitions, and handling of division by zero.

---

**1. Standard Division Operator (`/`)**

* **Python (Python 3+)**:
  * The `/` operator **always performs true (floating-point) division**, regardless of operand types.
  * Examples:
    ```python
    7 / 2    # -> 3.5 (float)
    6 / 2    # -> 3.0 (float)
    -7 / 2   # -> -3.5 (float)
    ```

* **Java**:
  * The behavior of `/` depends strictly on the **static types** of the operands:
    * **Integer division**: If both operands are integer types (`int`, `long`, `short`, `byte`), `/` truncates the decimal part (rounds toward zero).
    * **Floating-point division**: If at least one operand is a `float` or `double`, `/` performs floating-point division.
  * Examples:
    ```java
    7 / 2      // -> 3 (int)
    7 / 2.0    // -> 3.5 (double)
    (double) 7 / 2  // -> 3.5 (double)
    ```

---

**2. Integer Division and Rounding Direction**

The crucial difference lies in how integer division rounds, particularly for negative results:

| Concept | Python (`//`) | Java (`/` on ints) | Java (`Math.floorDiv`) |
| :--- | :--- | :--- | :--- |
| **Rounding Rule** | **Floor** (rounds down towards $-\infty$) | **Truncation** (rounds towards $0$) | **Floor** (rounds down towards $-\infty$) |
| `7 / 2` | `7 // 2` $\rightarrow$ `3` | `7 / 2` $\rightarrow$ `3` | `Math.floorDiv(7, 2)` $\rightarrow$ `3` |
| `-7 / 2` | `-7 // 2` $\rightarrow$ `-4` | `-7 / 2` $\rightarrow$ `-3` | `Math.floorDiv(-7, 2)` $\rightarrow$ `-4` |
| `7 / -2` | `7 // -2` $\rightarrow$ `-4` | `7 / -2` $\rightarrow$ `-3` | `Math.floorDiv(7, -2)` $\rightarrow$ `-4` |
| `-7 / -2`| `-7 // -2` $\rightarrow$ `3` | `-7 / -2` $\rightarrow$ `3` | `Math.floorDiv(-7, -2)` $\rightarrow$ `3` |

* **Python** also allows `//` on floating-point numbers: `10.5 // -3` evaluates to `-4.0`.

---

**3. Modulo / Remainder Operator (`%`)**

The modulo operator is tied directly to how integer division is computed ($a = q \times b + r$, so $r = a - q \times b$):

* **Python (`%` is true modulo)**:
  * Uses floor division $q = \lfloor a / b \rfloor$.
  * The result has the **same sign as the divisor** (or is zero).
  * Examples:
    ```python
    7 % 2    # -> 1
    -7 % 2   # -> 1   (since -7 = (-4 * 2) + 1)
    7 % -2   # -> -1  (since 7 = (-4 * -2) + -1)
    -7 % -2  # -> -1  (since -7 = (3 * -2) + -1)
    ```

* **Java (`%` is remainder / truncated division remainder)**:
  * Uses truncated division $q = \text{trunc}(a / b)$.
  * The result has the **same sign as the dividend** (or is zero).
  * Examples:
    ```java
    7 % 2    // -> 1
    -7 % 2   // -> -1  (since -7 = (-3 * 2) + -1)
    7 % -2   // -> 1   (since 7 = (-3 * -2) + 1)
    -7 % -2  // -> -1  (since -7 = (3 * -2) + -1)
    ```
  * *Note*: Java provides `Math.floorMod(a, b)` (since Java 8) to reproduce Python-style mathematical modulo.

---

**4. Division by Zero**

* **Python**:
  * Any division by zero (both integer and floating-point) raises an exception:
    ```python
    7 / 0    # Raises ZeroDivisionError: division by zero
    7.0 / 0  # Raises ZeroDivisionError: float division by zero
    ```

* **Java**:
  * **Integer division by zero** throws a runtime exception:
    ```java
    int x = 7 / 0; // Throws java.lang.ArithmeticException: / by zero
    ```
  * **Floating-point division by zero** follows IEEE 754 and does **not** throw an exception:
    ```java
    double a = 7.0 / 0.0;   // -> Double.POSITIVE_INFINITY
    double b = -7.0 / 0.0;  // -> Double.NEGATIVE_INFINITY
    double c = 0.0 / 0.0;   // -> Double.NaN
    ```

---

**Summary Checklist**

| Feature | Python | Java |
| :--- | :--- | :--- |
| **True Division** | `/` (always float) | `/` (only if at least one operand is float/double) |
| **Integer Division** | `//` (floor division) | `/` between integers (truncated toward 0) |
| **Python `//` Equivalent in Java** | `a // b` | `Math.floorDiv(a, b)` |
| **Sign of `%` Result** | Matches divisor | Matches dividend |
| **Python `%` Equivalent in Java** | `a % b` | `Math.floorMod(a, b)` |
| **Float Division by Zero** | Raises `ZeroDivisionError` | Returns `Infinity` or `NaN` |
